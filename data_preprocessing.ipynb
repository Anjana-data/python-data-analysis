```python
import re
```

```python
def price_clean(p):
    if '+' in p:
        v1=float(re.findall('\\d+',p)[0])
        v2=float(re.findall('\\d+',p)[1])
        return v1+(v2/100*v1)
    elif  '$' in p:
        v1=float(re.findall('(\\d+)\\$',p)[0])
        return v1*90
    else:
        return float(p)
```

```python
p_df['price']=p_df['price'].apply(price_clean).astype(int)
```

```python
p_df['price']
```

```python
p_df['discount'].unique()
```

```python
def disc_amt(r):
    d=r['discount'] #single cell value of the particular row
    p=r['price']
    if 'off' in d:
        n=float(re.findall('(\\d+)%\\s?off',d)[0])
        return p-(p*n/100) #assume 1000-(1000*20/100)
    elif 'flat' in d:
        n=float(re.findall('flat\\s?(\\d+)',d)[0])
        return (p-n)  #assume 1000 price -flat 300 =700
    elif 'upto' in d:
        n=float(re.findall('upto\\s?(\\d+)%',d)[0])//2 #upto 30% take half means 15?
        return p-(p*n/100)
    else:
        return float(p)


        
p_df['discounted amount']=p_df[['discount','price']].apply(disc_amt,axis=1) #axis=1 data will pass row by row
```
